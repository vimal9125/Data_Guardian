```python
# =========================
# 📘 DataGuardianAI.ipynb
# Machine Learning + AI Dataset Auditor (Jupyter Version)
# =========================

# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler
from datetime import datetime
import json

# Step 2: Load Dataset
# Replace with your dataset path or upload using Jupyter UI
df = pd.read_csv("your_dataset.csv")  # Example path
df.head()

# Step 3: Dataset Overview
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn Info:")
print(df.info())

# Step 4: Missing Values Summary
missing_summary = df.isnull().sum().reset_index()
missing_summary.columns = ['Column', 'Missing_Count']
missing_summary['Missing_%'] = (missing_summary['Missing_Count'] / len(df)) * 100
print("\nMissing Values Summary:")
print(missing_summary)

# Step 5: Basic Stats for Numeric Columns
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = [c for c in df.columns if c not in num_cols]

print("\nNumeric Columns:", num_cols)
print("Categorical Columns:", cat_cols)

numeric_stats = df[num_cols].describe().T
display(numeric_stats)

# Step 6: Anomaly Detection (Isolation Forest)
X = df[num_cols].fillna(df[num_cols].median())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

iso = IsolationForest(contamination=0.01, random_state=42)
df['_anomaly_if'] = iso.fit_predict(X_scaled)
df['_if_score'] = iso.decision_function(X_scaled)

print("\nIsolationForest Outliers:", (df['_anomaly_if'] == -1).sum())

# Step 7: (Optional) Local Outlier Factor
lof = LocalOutlierFactor(contamination=0.01)
df['_anomaly_lof'] = lof.fit_predict(X_scaled)
print("LOF Outliers:", (df['_anomaly_lof'] == -1).sum())

# Step 8: Visualize Anomalies
plt.figure(figsize=(8,4))
sns.countplot(x='_anomaly_if', data=df)
plt.title("Anomaly Distribution (IsolationForest)")
plt.show()

# Step 9: Missing Value Heatmap
plt.figure(figsize=(10,4))
sns.heatmap(df[num_cols].isna(), cbar=False)
plt.title("Missing Value Heatmap")
plt.show()

# Step 10: AI Summary (Mock)
ai_summary = f"""
# DataGuardianAI Audit Report
Generated: {datetime.utcnow().strftime('%Y-%m-%d %H:%M:%S UTC')}

**Rows:** {df.shape[0]}
**Columns:** {df.shape[1]}

- Total missing values detected: {int(missing_summary['Missing_Count'].sum())}
- Outliers (IsolationForest): {(df['_anomaly_if'] == -1).sum()}
- Outliers (LOF): {(df['_anomaly_lof'] == -1).sum()}

💡 **AI Suggestions (example):**
- Impute missing values with median for numeric columns.
- Drop columns having >50% missing data.
- Review flagged rows with anomaly = -1.
"""

print(ai_summary)

# Step 11: Save Final Report
with open("DataGuardianAI_Report.md", "w") as f:
    f.write(ai_summary)

print("✅ Report saved as DataGuardianAI_Report.md")
```
